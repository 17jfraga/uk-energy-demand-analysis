```python
import pandas as pd

filename = next(iter(uploaded))

df = pd.read_csv(filename)

print("Dataset loaded successfully!")
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))
```

```python
print("First Five Rows")
display(df.head())

print("Column Names")
print(df.columns.tolist())

print("Summary Statistics")
print(df.isna().sum())

print("Unique Values")
print(df.nunique())
```

```python
def summarise_column(data, column):
  """Print basic statistics for one numerical column."""
  values = data[column].dropna()

  print(f"\nColumn: {column}")
  print(f"Average: {values.mean():,.2f}")
  print(f"Minimum: {values.min():,.2f}")
  print(f"Maximum: {values.max():,.2f}")

numeric_columns = df.select_dtypes(
    include="number"
).columns

for column in numeric_columns:
  summarise_column(df, column)
```

```python
df["SETTLEMENT_DATE"] = pd.to_datetime(df["SETTLEMENT_DATE"])

print("Number of observations:", len(df))
print("Number of columns;", len(df.columns))
print("First date:", df["SETTLEMENT_DATE"].min())
print("Last date:", df["SETTLEMENT_DATE"].max())

print("\nMissing values by column:")
print(df.isna().sum())

key_columns = [
    "ND",
    "TSD",
    "ENGLAND_WALES_DEMAND",
    "EMBEDDED_WIND_GENERATION",
    "EMBEDDED_SOLAR_GENERATION"
]

display(df[key_columns].describe().round(2))
```

```python
def analyse_measure(data, column):
    values = data[column].dropna()

    print(f"\nAnalysis of {column}")
    print(f"Mean: {values.mean():,.1f} MW")
    print(f"Median: {values.median():,.1f} MW")
    print(f"Minimum: {values.min():,.1f} MW")
    print(f"Maximum: {values.max():,.1f} MW")


analyse_measure(df, "ND")
analyse_measure(df, "EMBEDDED_WIND_GENERATION")
analyse_measure(df, "EMBEDDED_SOLAR_GENERATION")
```

```python
import matplotlib.pyplot as plt

daily = df.groupby("SETTLEMENT_DATE")[
    [
        "ND",
        "EMBEDDED_WIND_GENERATION",
        "EMBEDDED_SOLAR_GENERATION"
    ]
].mean()

daily["ND"].plot(figsize=(12,5))

plt.title("Average Half-hourly National Demand by Day -2025")
plt.xlabel("Date")
plt.ylabel("Average demand (MW)")
plt.grid(True)
plt.show()
```

```python
daily["EMBEDDED_SOLAR_GENERATION"].plot(figsize=(12, 5))

plt.title("Average Daily Embedded Solar Generation — 2025")
plt.xlabel("Date")
plt.ylabel("Average generation (MW)")
plt.grid(True)
plt.show()
```

Initial findings — UK electricity data, 2025
1. National demand averaged approximately 26.2 GW, with a recorded range of 12.8–45.9 GW across settlement periods.
2. Embedded wind generation averaged approximately 1.95 GW and varied considerably over the year.
3. Embedded solar generation averaged approximately 2.13 GW, but its median was just 7 MW, suggesting a highly uneven distribution across settlement periods.
4. The dataset contains 17,520 observations, consistent with 48 half-hourly periods per day across 2025.
5. These results motivate further investigation into the timing and reliability of electricity supply for energy-intensive infrastructure, including data centres.

```python
clean_df = df.copy()

clean_df["SETTLEMENT_DATE"] = pd.to_datetime(clean_df["SETTLEMENT_DATE"], errors="coerce")

numeric_cols = ["SETTLEMENT_PERIOD",
                "ND",
                "EMBEDDED_WIND_GENERATION",
                "EMBEDDED_SOLAR_GENERATION"
]

for col in numeric_cols:
  clean_df[col] = pd.to_numeric(clean_df[col], errors="coerce")

print("Missing values before cleaning:")
print(clean_df.isna().sum().sort_values(ascending=False))

print("Duplicate rows:", clean_df.duplicated().sum())

clean_df = clean_df.drop_duplicates()

print("Cleaned dataset shate:", clean_df.shape)



```

```python
high_demand = clean_df[clean_df["ND"]>40000]
print("High-demand observations:", len(high_demand))

display(
    clean_df.sort_values("ND", ascending=False).head(10)
)

print("Average demand:", clean_df["ND"].mean())
print("Maximum demand:", clean_df["ND"].max())
print("Minimum demand:", clean_df["ND"].min())

```

```python
average_demand = (
    clean_df
    .dropna(subset=["SETTLEMENT_PERIOD", "ND"])
    .groupby("SETTLEMENT_PERIOD")["ND"]
    .mean()
    .reset_index()
)

average_demand.columns = [
    "SETTLEMENT_PERIOD",
    "AVERAGE_DEMAND"
]

display(average_demand.head())
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    average_demand["SETTLEMENT_PERIOD"],
    average_demand["AVERAGE_DEMAND"]
)

plt.title("Average UK National Electricity Demand by Settlement Period")
plt.xlabel("Settlement period (half-hourly intervals)")
plt.ylabel("Average national demand (MW)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
solar_by_period = (
    clean_df
    .dropna(subset=[
        "SETTLEMENT_PERIOD",
        "EMBEDDED_SOLAR_GENERATION"
    ])
    .groupby("SETTLEMENT_PERIOD")["EMBEDDED_SOLAR_GENERATION"]
    .mean()
)

solar_by_period.plot(
    figsize=(12,5),
    title="Average Embedded Solar Generation by Settlement Period"
)

plt.xlabel("Settlement period")
plt.ylabel("Average embedded solar generation (MW)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
df["SETTLEMENT_DATE"] = pd.to_datetime(df["SETTLEMENT_DATE"])

df["HOUR"] = (df["SETTLEMENT_PERIOD"]-1) //2

daily = df.groupby("SETTLEMENT_DATE")["ND"].mean()
```

```python
plt.figure(figsize=(12,5))

plt.plot(daily.index, daily.values)

plt.title("Average Daliy Demand in 2025")
plt.xlabel("Date")
plt.ylabel("Average Demand (MW)")

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
monthly = df.groupby(
    pd.Grouper(key="SETTLEMENT_DATE", freq="ME")
)[["EMBEDDED_WIND_GENERATION",
   "EMBEDDED_SOLAR_GENERATION"]].mean()

plt.figure(figsize=(12,5))

plt.plot(monthly.index, monthly["EMBEDDED_WIND_GENERATION"],
         label="Embedded Wind")

plt.plot(monthly.index, monthly["EMBEDDED_SOLAR_GENERATION"],
         label="Embedded Solar")

plt.title("Average Monthly Embedded Wind and Solar Generation")
plt.xlabel("Month")
plt.ylabel("Average Generation (MW)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
hourly = df.groupby("HOUR")["ND"].mean()

plt.figure(figsize=(10,5))

plt.plot(hourly.index, hourly.values, marker="o")

plt.title("Average Electricity Demand by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Average Demand (MW)")

plt.xticks(range(24))
plt.grid(True)
plt.tight_layout()
plt.show()
```
